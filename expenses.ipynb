```python

import database
```

```python

USER_CHOICE = """
Enter:

- 'a' to add a new expense
- 'l' to list all expenses
- 's' to search for a expense
- 'd' to delete a expense
- 'p' to update expense amount
- 'q' to quit

Your choice: """
```

```python

def prompt_add_expense():

    title = input("Enter expense title: ")
    category = input("Enter expense category: ")
    amount = float(input("Enter expense amount: "))
    date = input("Enter expense date: ")
    
    database.insert_expense(title,category,amount,date)

```

```python

def list_expenses():

  expenses = database.get_all_expenses()

  for expense in expenses:
    print(
            f"{expense[0]} - {expense[1]} "
            f"| Amount: Rs{expense[2]} "
            f"| Date: {expense[3]}"
        )

```

```python

def prompt_search_expense():
    title = input("Enter expense title: ")
    expense = database.search_expense(title)

    if expense:
        print("\nExpense Found!")
        print(f"Title   : {expense[0]}")
        print(f"Category : {expense[1]}")
        print(f"Amount   : Rs{expense[2]}")
        print(f"Date  : {expense[3]}")
    else:
        print("Expense not found!")
```

```python

def prompt_update_expense_amount():
    title = input("Enter expense title: ")
    new_amount = float(input("Enter new amount: "))

    database.update_expense_amount(title, new_amount)
    
```

```python
def menu():

    user_input = input(USER_CHOICE)
    
    while user_input != "q":
        
        if user_input == "a":
            prompt_add_expense()
        
        elif user_input == "l":
            list_expenses()
        
        elif user_input == "s":
            prompt_search_expense()
        
        elif user_input == "d":
            prompt_delete_expense()
        
        elif user_input == "p":
            prompt_update_expense_amount()
        
        else:
            print("Invalid choice!")

        user_input = input(USER_CHOICE)
```

```python
menu()
```
